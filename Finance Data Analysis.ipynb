```python
data=data.drop(["Experience"], axis=1)
```

```python
# education
data["Education"].unique()
```

```python
def ed(x):
    if x==1:
        return "Undergrade"
    if x==2:
        return "Graduate"
    if x==3:
        return "Professional person"
```

```python
data["Edu"]=data["Education"].apply(ed)
```

```python
data["Securities_Account"].value_counts()
```

```python
def security(y):
    if (y["Securities_Account"]==1) &(y["CD_Account"]==1):
        return "Hold Securities & Deposite"
    if (y["Securities_Account"]==0) &(y["CD_Account"]==0):
        return "Does not Hold Securities & Deposite account"
    if (y["Securities_Account"]==1) &(y["CD_Account"]==0):
        return "Hold only Securities account"
    if (y["Securities_Account"]==0) &(y["CD_Account"]==1):
        return "Hold only Deposite account"
```

```python
data["Account_holder_category"]= data.apply(security, axis=1)
```

```python
data.head()
```

```python
values=data["Account_holder_category"].value_counts()
values.index
```

```python
plt.pie(x=values.values,labels=values.index,autopct='%0.0f%%',radius=0.9)
# plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data,x="Edu",y="Income",hue="Personal_Loan")
```

```python
plt.figure(figsize=(12,6))
sns.distplot(data[data["Personal_Loan"]==0]["Income"],label="Income with no personal loan")

sns.distplot(data[data["Personal_Loan"]==1]["Income"],label="Income with personal loan")

plt.legend()
```

```python
def plot(col1,col2,label1,label2,title):
    plt.figure(figsize=(12,6))
    sns.distplot(data[data[col2]==0][col1],label=label1,hist=False)
    
    sns.distplot(data[data[col2]==1][col1],label=label2,hist=False)
    
    plt.legend()
    plt.title(title)
```

```python
plot("Income","Personal_Loan","Income with no personal loan","Income with personal loan","Income Distribution plot")
```

```python
plot("CCAvg","Personal_Loan","Credit card avg with no personal loan","Credit card avg with personal loan",
     "Credit card avg Distribution")
```
